## Income Distribution

Understanding customer income helps identify purchasing power segments.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df_clean["Income"], bins=30)
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

### Insight

Income distribution is concentrated in mid-range values, with a few high-income outliers.  
This suggests a heterogeneous customer base with distinct purchasing power segments.

## Age Distribution

```python
sns.histplot(df_clean["Age"], bins=30)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### Insight

Most customers are middle-aged, which is typical for long-term consumer goods behavior.

## Total Spending Distribution

```python
sns.histplot(df_clean["MntTotal"], bins=30)
plt.title("Total Spending Distribution")
plt.xlabel("Total Spending")
plt.ylabel("Frequency")
plt.show()
```

### Insight

Spending behavior is highly skewed, with a small group of high-value customers driving most of the consumption.

## Income vs Total Spending

```python
sns.scatterplot(data=df_clean, x="Income", y="MntTotal")
plt.title("Income vs Total Spending")
plt.show()
```

### Insight

There is a positive relationship between income and total spending, although with noticeable dispersion, suggesting other behavioral factors influence purchasing decisions.

## Correlation Heatmap

This helps identify relationships between numeric variables.

```python
plt.figure(figsize=(12,8))
sns.heatmap(df_clean.corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

### Insight

Strong correlations are expected between total spending and product-specific spending categories.
